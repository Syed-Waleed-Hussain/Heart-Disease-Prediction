# Heart Disease Prediction

Import Libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

Import Dataset

```python
df = pd.read_csv('/content/cleaned_merged_heart_dataset.csv')
```

Splitting the dataset into features and target columns

```python
x = df.drop(columns = 'target', axis = 1)
y = df['target']
```

```python
print(x)
```

```python
print(y)
```

Splittng the dataset into train and test dataset

```python
x_train,x_test,y_train,y_test = train_test_split(x , y , test_size = 0.2 , stratify=y , random_state=2)
```

```python
print(x.shape, x_train.shape, x_test.shape)
```

```python
print(y.shape,y_train.shape, y_test.shape)
```

Training The model "Logistic Regression

```python
model = LogisticRegression()
```

```python
model.fit(x_train , y_train)
```

Model Evaluation

Accuracy on training data

```python
x_train_prediction = model.predict(x_train)
train_accuracy = accuracy_score(x_train_prediction , y_train)

print("Accuracy on X_train dataset is :", train_accuracy)
```

Accuracy score on Test dataset

```python
x_test_prediction = model.predict(x_test)
test_accuracy = accuracy_score(x_test_prediction , y_test)

print("Accuracy on X_test dataset is :", test_accuracy)
```

Building Predictive System

```python
input_data = (52,1,1,120,325,0,1,172,0,0.2,2,0,2)
input_data_into_numpy_array = np.asarray(input_data)
input_data_reshape = input_data_into_numpy_array.reshape(1,-1)

prediction = model.predict(input_data_reshape)
print(prediction)

if (prediction[0] == 0):
  print("Patient's heart is healthy")
else:
  print("Patient has heart disease")
```
